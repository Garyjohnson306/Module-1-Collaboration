```python
class Solution:
    def sort012(self, arr):
        
        zeros = 0
        ones = 0
        twos = 0
        
        for number in arr:
            
            if number == 0:
                zeros += 1
            elif number == 1:
                ones += 1
            else:
                twos += 1
        
        for i in range(len(arr)):
            
            if i < zeros:
                arr[i] = 0
            elif i < zeros + ones:
                arr[i] = 1
            else:
                arr[i] = 2
                
test = [0, 0, 0, 1, 1, 1, 1, 2, 2, 1, 0, 1, 0, 2, 2, 1, 2, 0]
solution = Solution()
solution.sort012(test)

print(test)
```

```python
class Solution:
    def sort012(self, arr):

        low = 0
        mid = 0
        high = len(arr) - 1

        while mid <= high:
            if arr[mid] == 0:
                arr[low], arr[mid] = arr[mid], arr[low]
                low += 1
                mid += 1
            elif arr[mid] == 1:
                mid += 1
            else:
                arr[mid], arr[high] = arr[high], arr[mid]
                high -= 1

test = [1, 1, 0, 1, 2, 0, 2, 1, 2, 2, 2, 1, 0, 0]
solution = Solution()
solution.sort012(test)

print(test)
```

```python
class Solution:
    def firstSearch(self, arr, k):

        low = 0
        high = len(arr) - 1
        result = -1

        while low <= high:
            mid = (low + high) // 2

            if arr[mid] == k:
                result = mid
                high = mid - 1
            elif arr[mid] < k:
                low = mid + 1
            else:
                high = mid - 1
        return result

test = [1, 2, 2, 3, 4, 5, 7, 9,]
k = 2

solution = Solution()
result = solution.firstSearch(test, k)

print(result)

test = [1, 1, 3, 4, 7, 8]
k= 10

solution = Solution()
result = solution.firstSearch(test, k)

print(result)
```
